```python
import yt_dlp

def get_video_details(url):
    options = {
        'quiet': True,
        'no_warnings': True
    }

    try:
        with yt_dlp.YoutubeDL(options) as ydl:
            info = ydl.extract_info(url, download=False)

            print("Video Title:", info.get('title'))
            print("Channel Name:", info.get('uploader'))
            print("Duration:", info.get('duration'), "seconds")
            print("Video URL:", url)

    except Exception as e:
        print("Error:", e)

print("YouTube Video Details Extractor")
```

```python
video_url = input("Enter YouTube Video URL: ")

get_video_details(video_url)
```

```python
import os
import yt_dlp

os.environ["PATH"] += os.pathsep + "/root/.deno/bin"

def download_video(url):
    options = {
        'format': 'best',
        'outtmpl': '/content/%(title)s.%(ext)s',
        'noplaylist': True,
        'js_runtimes': {'deno': {}},
        'remote_components': {'ejs:github'}
    }

    try:
        with yt_dlp.YoutubeDL(options) as ydl:
            ydl.download([url])

        print("Video download completed!")

    except Exception as e:
        print("Download error:", e)

print("YouTube Downloader Ready!")
```

```python
video_url = input("Enter YouTube Video URL: ")

download_video(video_url)
```
